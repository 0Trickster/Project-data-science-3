# Ejecución de Modelo Random Forest desde Archivo PKL

Este notebook tiene como objetivo cargar un modelo de clasificación Random Forest previamente entrenado y almacenado en formato .pkl, para posteriormente evaluar su rendimiento utilizando el conjunto de datos de prueba.

El modelo fue entrenado utilizando el conjunto de datos procesado del proyecto Student Depression Dataset.

## Objetivo

### Validar el desempeño del modelo entrenado mediante:

* Carga de los conjuntos de entrenamiento y prueba procesados.
* Recuperación del modelo serializado.
* Generación de predicciones sobre el conjunto de prueba.
* Evaluación mediante métricas de clasificación.

## Librerías utilizadas

```python
import pandas as pd
import joblib
from src.carga_csv import cargar_csv
```

| Librería | Propósito |
|-----------|-----------|
| pandas | Manipulación y carga de datos |
| joblib | Serialización y carga de modelos |
| scikit-learn | Métricas de evaluación |

```python
model_rf = joblib.load(
    "../outputs/models/random_forest_model.pkl"
)

print("Modelo cargado correctamente.")
```

```python
preprocessor = joblib.load(
    r'..\outputs\models\preprocessor.pkl'
)
```

## Carga de Datos

Se cargan los conjuntos de entrenamiento y prueba previamente transformados durante la etapa de preprocesamiento.


```python
df = cargar_csv(
    r"..\data\processed\Student_Depression_Dataset_Prediction.csv"
)

df.head()
```

```python
X_nuevo = preprocessor.transform(df)

X_nuevo = pd.DataFrame(
    X_nuevo,
    columns=preprocessor.get_feature_names_out()
)
```

## Generación de Predicciones

Una vez cargado el modelo, se generan predicciones sobre el conjunto de prueba.


```python
predicciones = model_rf.predict(X_nuevo)
```

Cada predicción corresponde a una de las siguientes clases:

| Valor | Interpretación |
|---------|--------------|
| 0 | No presenta depresión |
| 1 | Presenta depresión |

```python
predicciones
```

```python
probabilidades = model_rf.predict_proba(X_nuevo)

probabilidades
```

```python
resultado = df.copy()

resultado["Prediccion"] = predicciones
resultado["Prob_Depression"] = (
    model_rf.predict_proba(X_nuevo)[:, 1]
)

resultado
```
